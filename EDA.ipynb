```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('titanic_data_updated.csv')
```

```python
df.shape
print(f"Total potential feature : {df.shape[1]}")
print(f"Total sample data : {df.shape[0]}")
```

```python
df.head(10)
```

```python
# Random
df.sample(10)
```

```python
#missing
df.isnull().sum()
```

```python
#checking duplicates
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

```python
df.info()
```

```python
#statistical
df.describe()
```

Univariate Analysis (Categorical Value Analysis -1)

```python
sns.countplot(data=df, x='Survived')
```

```python
(df['Survived'].value_counts() / len(df)) * 100
```

```python
survived_count = df['Survived'].value_counts()
survived_label = df['Survived'].unique()
print(survived_label)
plt.pie(survived_count,labels = survived_label, autopct="%1.1f")
plt.title("Pie Chart of Survived")
plt.show()
```

Univariate Analysis (Numerical Column Analysis - 1)

```python
sns.histplot(df['Age'], bins = 50)
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.kdeplot(df['Age'])
plt.title("Density Plot of Age")
plt.xlabel("Age")
plt.ylabel("Probability Distribution")
plt.show()
```

```python
sns.histplot(df['Fare'], bins = 50)
plt.title("Plot of Fare")
plt.xlabel("Fare")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.kdeplot(df['Fare'])
plt.title("Plot of Fare")
plt.xlabel("Fare")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.boxplot(df['Fare'])
```

Multivariate Analysis- 1 ( Countplot with Hue-1)

```python
sns.countplot(x = df['Sex'], hue = df['Survived'])
```

```python
gender_split = df.groupby('Sex')['Survived'].value_counts(normalize=True)
print(gender_split)
```

```python
sns.countplot(x = df['Pclass'], hue = df['Survived'])
```

```python
pclass_split = df.groupby('Pclass')['Survived'].value_counts(normalize=True)
print(pclass_split)
```

Barplot for Cat-Num

```python
sns.barplot(x = df['Pclass'], y = df['Fare'])
```

```python
sns.barplot(x = df['Survived'], y = df['Age'], hue = df['Sex'])
```
